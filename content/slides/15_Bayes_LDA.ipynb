## Bayes Decision Theory & The Optimal Bayes Classifier

In supervised classification, our goal is to predict a discrete class label $Y \in \{1, 2, \dots, C\}$ for an observed feature vector $X = (x_1, x_2, \dots, x_d)^T \in \mathbb{R}^d$.

### Bayes' Theorem for Classification

Bayes' Theorem converts conditional feature likelihoods into class posterior probabilities:

$$P(Y=c \mid X=x) = \frac{P(X=x \mid Y=c) P(Y=c)}{P(X=x)}$$

where:

* **$P(Y=c)$ ($\pi_c$):** The **prior probability** of class $c$ (the baseline prevalence of the class before observing features).
* **$P(X=x \mid Y=c)$:** The **class-conditional probability density** (likelihood) of observing features $x$ given class $c$.
* **$P(X=x)$:** The **marginal likelihood** across all classes:

$$P(X=x) = \sum_{j=1}^{C} P(X=x \mid Y=j) P(Y=j)$$


* **$P(Y=c \mid X=x)$:** The **posterior probability** that feature vector $x$ belongs to class $c$.

---

To minimize the expected misclassification rate (0-1 loss), the **Bayes Decision Rule** assigns an observation $x$ to the class with the maximum posterior probability:

$$\hat{y} = \arg\max_{c \in \{1, \dots, C\}} P(Y=c \mid X=x)$$

Because the marginal likelihood $P(X=x)$ is constant for all classes $c$ for a given $x$, maximizing the posterior is equivalent to maximizing the joint probability:

$$\hat{y} = \arg\max_{c \in \{1, \dots, C\}} P(X=x \mid Y=c) P(Y=c)$$

> The optimal Bayes classifier represents the theoretical minimum bound for misclassification error on a given data distribution. Any actual classifier strives to approximate the true probability densities $P(X \mid Y)$ and priors $P(Y)$.

---

## Linear Discriminant Analysis (LDA)

Linear Discriminant Analysis (LDA) is a **generative model** that assumes the class-conditional distributions $P(X \mid Y=c)$ follow a **Multivariate Normal (Gaussian) distribution** with a **shared covariance matrix** across all classes.

### Assumptions

* **Gaussian Distribution:** Features for class $c$ are distributed as multivariate normal:

$$X \mid (Y=c) \sim \mathcal{N}(\mu_c, \Sigma)$$



where $\mu_c \in \mathbb{R}^d$ is the mean vector for class $c$.
* **Equal Covariance:** All $C$ classes share the exact same covariance matrix $\Sigma$:

$$\Sigma_1 = \Sigma_2 = \dots = \Sigma_C = \Sigma$$


The multivariate Gaussian density function for class $c$ is:

$$P(X=x \mid Y=c) = \frac{1}{(2\pi)^{p/2} \vert{}\Sigma\vert{}^{1/2}} \exp\left( -\frac{1}{2} (x - \mu_c)^T \Sigma^{-1} (x - \mu_c) \right)$$

> $p$ is the number of features

---

### Derivation of the Linear Discriminant Function

To classify point $x$, we take the natural logarithm of the joint density $P(X=x \mid Y=c) \cdot \pi_c$:

$$\ln \left( P(X=x \mid Y=c) \pi_c \right) = \ln P(X=x \mid Y=c) + \ln \pi_c$$

Substitute the multivariate Gaussian density:

$$\ln \left( P(X=x \mid Y=c) \pi_c \right) = -\frac{d}{2}\ln(2\pi) - \frac{1}{2}\ln\vert{}\Sigma\vert{} - \frac{1}{2}(x - \mu_c)^T \Sigma^{-1} (x - \mu_c) + \ln \pi_c$$

Expanding the quadratic term $(x - \mu_c)^T \Sigma^{-1} (x - \mu_c)$:

$$(x - \mu_c)^T \Sigma^{-1} (x - \mu_c) = x^T \Sigma^{-1} x - 2 x^T \Sigma^{-1} \mu_c + \mu_c^T \Sigma^{-1} \mu_c$$

Now drop all terms that do not depend on the class index $c$ ($-\frac{d}{2}\ln(2\pi)$, $-\frac{1}{2}\ln\vert{}\Sigma\vert{}$, and $-\frac{1}{2} x^T \Sigma^{-1} x$):

$$\delta_c(x) = x^T \Sigma^{-1} \mu_c - \frac{1}{2} \mu_c^T \Sigma^{-1} \mu_c + \ln \pi_c$$

$\delta_c(x)$ is the **Linear Discriminant Function** for class $c$.

The classification decision rule simplifies to:

$$\hat{y} = \arg\max_{c} \delta_c(x)$$

Notice that $\delta_c(x)$ is **linear** with respect to $x$. The quadratic term $x^T \Sigma^{-1} x$ canceled out specifically because $\Sigma$ is identical for all classes.

---

### Binary Decision Boundary

For a binary classification problem ($C = 1, 2$), the decision boundary occurs where $\delta_1(x) = \delta_2(x)$:

$$x^T \Sigma^{-1} \mu_1 - \frac{1}{2} \mu_1^T \Sigma^{-1} \mu_1 + \ln \pi_1 = x^T \Sigma^{-1} \mu_2 - \frac{1}{2} \mu_2^T \Sigma^{-1} \mu_2 + \ln \pi_2$$

Rearranging into linear hyperplane form $w^T x + b = 0$:

$$w^T x + b = 0$$

where:

* **Weight vector:** $w = \Sigma^{-1} (\mu_1 - \mu_2)$
* **Bias term:** $b = -\frac{1}{2} \mu_1^T \Sigma^{-1} \mu_1 + \frac{1}{2} \mu_2^T \Sigma^{-1} \mu_2 + \ln \left( \frac{\pi_1}{\pi_2} \right)$

---

## Parameter Estimation from Training Data

In practice, population parameters ($\mu_c, \Sigma, \pi_c$) are unknown and must be estimated from $N$ training samples with class counts $N_c$:

1. **Prior Probabilities:**

$$\hat{\pi}_c = \frac{N_c}{N}$$


2. **Class Sample Means:**

$$\hat{\mu}_c = \frac{1}{N_c} \sum_{i: y_i = c} x_i$$


3. **Pooled Covariance Matrix:**

$$\hat{\Sigma} = \frac{1}{N - C} \sum_{c=1}^{C} \sum_{i: y_i = c} (x_i - \hat{\mu}_c)(x_i - \hat{\mu}_c)^T$$

## Worked Example

### Problem Statement

Given a 1D dataset ($d = 1$) with two classes $C_1$ and $C_2$:

* **Class 1 ($C_1$):** Mean $\mu_1 = 3$, Variance $\sigma^2 = 4$, Prior $\pi_1 = 0.8$
* **Class 2 ($C_2$):** Mean $\mu_2 = 7$, Variance $\sigma^2 = 4$, Prior $\pi_2 = 0.2$

Find the optimal decision boundary $x^*$ and classify a query point $x_{\text{test}} = 5.5$.

---

#### 1: Discriminant Functions ($\delta_1(x)$ and $\delta_2(x)$)

For 1D data, $\Sigma^{-1} = \frac{1}{\sigma^2}$. The discriminant formula becomes:

$$\delta_c(x) = x \cdot \frac{\mu_c}{\sigma^2} - \frac{\mu_c^2}{2\sigma^2} + \ln \pi_c$$

* **For Class 1:**

$$\delta_1(x) = x \cdot \frac{3}{4} - \frac{3^2}{2(4)} + \ln(0.8) = 0.75x - \frac{9}{8} + \ln(0.8)$$


$$\delta_1(x) \approx 0.75x - 1.125 - 0.2231 = 0.75x - 1.3481$$


* **For Class 2:**

$$\delta_2(x) = x \cdot \frac{7}{4} - \frac{7^2}{2(4)} + \ln(0.2) = 1.75x - \frac{49}{8} + \ln(0.2)$$


$$\delta_2(x) \approx 1.75x - 6.125 - 1.6094 = 1.75x - 7.7344$$



---

#### 2: Decision Boundary ($x^*$)

Set $\delta_1(x^*) = \delta_2(x^*)$:

$$0.75 x^* - 1.3481 = 1.75 x^* - 7.7344$$

$$1.00 x^* = 6.3863 \implies x^* \approx 6.386$$

> **Note:** If the priors were equal ($\pi_1 = \pi_2 = 0.5$), the decision boundary would be exactly halfway between the means: $\frac{\mu_1 + \mu_2}{2} = \frac{3 + 7}{2} = 5.0$. Because $\pi_1 = 0.8 > \pi_2 = 0.2$, the boundary shifts to the right ($6.386$), requiring stronger evidence to declare Class 2.

---

#### 3: Classify Query $x_{\text{test}} = 5.5$

* $\delta_1(5.5) = 0.75(5.5) - 1.3481 = 4.125 - 1.3481 = \mathbf{2.7769}$
* $\delta_2(5.5) = 1.75(5.5) - 7.7344 = 9.625 - 7.7344 = \mathbf{1.8906}$

Since $\delta_1(5.5) > \delta_2(5.5)$, the query point $x = 5.5$ is classified as **Class 1 ($C_1$)**.

## Python Implementation

```
import numpy as np
from sklearn.datasets import make_classification
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


# Dataset Generation & Splitting
X, y = make_classification(
    n_samples=500,
    n_features=5,
    n_informative=2,
    n_classes=2,
    weights=[0.6, 0.4],
    random_state=42,
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# -------------------------------------------------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -------------------------------------------------------------------
classes = np.unique(y_train)
n_samples, n_features = X_train_scaled.shape

# Compute Priors and Class Means
priors = {}
means = {}
for c in classes:
    X_c = X_train_scaled[y_train == c]
    priors[c] = X_c.shape[0] / n_samples
    means[c] = X_c.mean(axis=0)

# Compute Pooled Covariance Matrix
pooled_cov = np.zeros((n_features, n_features))
for c in classes:
    X_c = X_train_scaled[y_train == c]
    diff = X_c - means[c]
    pooled_cov += np.dot(diff.T, diff)

pooled_cov /= n_samples - len(classes)
inv_cov = np.linalg.inv(pooled_cov)

```

```
def predict_lda_manual(X_data):
    predictions = []
    for x in X_data:
        scores = {}
        for c in classes:
            # delta_c(x) = x^T * Sigma^-1 * mu_c - 0.5 * mu_c^T * Sigma^-1 * mu_c + ln(pi_c)
            term1 = np.dot(x.T, np.dot(inv_cov, means[c]))
            term2 = 0.5 * np.dot(means[c].T, np.dot(inv_cov, means[c]))
            term3 = np.log(priors[c])
            scores[c] = term1 - term2 + term3
        predictions.append(max(scores, key=scores.get))
    return np.array(predictions)


y_pred_manual = predict_lda_manual(X_test_scaled)
print(f" LDA Test Accuracy: {accuracy_score(y_test, y_pred_manual):.2f}")


```

```
# Scikit-Learn LDA Implementation

lda_model = LinearDiscriminantAnalysis()
lda_model.fit(X_train_scaled, y_train)

y_pred_sklearn = lda_model.predict(X_test_scaled)
print(
    f"Scikit-Learn LDA Test Accuracy: {accuracy_score(y_test, y_pred_sklearn):.2f}"
)

print("\nClassification Report:\n", classification_report(y_test, y_pred_sklearn))
```


### LDA is Generative

* Models the Joint Probability: Instead of directly learning the class boundary $(P(Y\vert{}X))$, LDA models the joint probability $(P(X, Y))$ by breaking it down into the prior probability of each class $(P(Y))$ and the class-conditional probability (likelihood) $(P(X\vert{}Y))$.
* Assumes a Data Distribution: LDA assumes that the features for each class follow a Gaussian (normal) distribution with a shared covariance matrix.
* Uses Bayes' Theorem: To classify a new data point, it plugs these estimated distributions into Bayes' theorem to calculate the probability of each class.
* Can Generate New Samples: Because it learns the underlying statistical distribution $((P(X\vert{}Y)))$ of the features for each category, you can mathematically sample from those Gaussian distributions to "generate" brand-new, synthetic data points that look like the training data.

* The name 'Discriminant' comes from its purpose. Even though the underlying mechanism builds a generative probabilistic model, the task it is designed to solve is discriminating (separating) between different classes or categories.
